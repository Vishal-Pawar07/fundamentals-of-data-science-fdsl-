```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
import pandas as pd
```

```python
df=pd.read_csv('student_test_score.csv')
data=df['exam_score']
```

```python
mean=np.mean(data)
std_dev=np.std(data)
print("Mean :",mean)
print("standard deviation:",std_dev)
```

```python
x=np.linspace(mean-4*std_dev,mean+4*std_dev,1000)
y=norm.pdf(x,mean,std_dev)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(x, y, color='blue', label='Normal Distribution Curve')
plt.hist(data, bins=30, density=True, alpha=0.4, color='orange', label='Actual Data')
plt.axvline(mean, color='red', linestyle='--', label='Mean')
plt.axvline(mean + std_dev, color='green', linestyle=':', label='+1 Std Dev')
plt.axvline(mean - std_dev, color='green', linestyle=':', label='-1 Std Dev')
plt.title('Normal Distribution Curve')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```
